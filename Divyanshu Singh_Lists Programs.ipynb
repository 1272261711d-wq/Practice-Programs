```python
#Accept 20 values from user and save it in list. Perform following operations on it
numbers = []

for i in range(20):
    value = int(input("Enter value: "))
    numbers.append(value)

print("\nList:", numbers)

#a)Count similar elements of list and print their index value
print("\nSimilar elements and their index values:")

for value in numbers:
    if numbers.count(value) > 1:
        indexes = []
        for i in range(len(numbers)):
            if numbers[i] == value:
                indexes.append(i)
print(value, "occurs", numbers.count(value), "times at indexes", indexes)

#b)Sort list in descending order using sort() function
even = 0
odd = 0

for value in numbers:
    if value % 2 == 0:
        even += 1
    else:
        odd += 1

print("\nEven values:", even)
print("Odd values:", odd)

#c)Count positive and negative values of list
positive = 0
negative = 0

for value in numbers:
    if value > 0:
        positive += 1
    elif value < 0:
        negative += 1

print("Positive values:", positive)
print("Negative values:", negative)
```

```python
#Accept 10 values from user and save it in list. Perform following operations on it
numbers = []

for i in range(10):
    value = int(input("Enter value: "))
    numbers.append(value)

#a)Sort list in ascending order using sorted() function and display sorted list
ascending = sorted(numbers)
print("\nAscending order:", ascending)

#Sort list in descending order using sort() function
numbers.sort(reverse=True)
print("Descending order:", numbers)

#c) display length of list
print("Length of list:", len(numbers))
```

```python
#Accept two lists from user and merge them using '+' in a single list
list1 = []
list2 = []

n1 = int(input("Enter number of elements in list 1: "))

for i in range(n1):
    value = int(input("Enter element: "))
    list1.append(value)

n2 = int(input("Enter number of elements in list 2: "))

for i in range(n2):
    value = int(input("Enter element: "))
    list2.append(value)

merged_list = list1 + list2

print("List 1:", list1)
print("List 2:", list2)
print("Merged list:", merged_list)
```

```python
#An acronym is a word formed by taking the first letters of the words in a phrase and making a word from them. For example, RAM is an acronym for “random access memory.” Write a program that allows the user to type in a phrase and then outputs the acronym for that phrase. Note the acronym should be all uppercase, even if the words in the phrase are not capitalized
phrase = input("Enter a phrase: ")
words = phrase.split() #(Split() separates the phrase into individual parts

acronym = ""

for word in words:
    acronym = acronym + word[0]

print("Acronym:", acronym.upper())
```

```python
#Write a program to print the abbreviation of a month, given its number
month = int(input("Enter month number: "))

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

if 1 <= month <= 12:
    print("Month abbreviation:", months[month - 1]) #(User will use [month - 1] as list indexing starts from 0, while month numbers start from 1)
else:
    print("Invalid month number")
```

```python
### Write a Python program to insert, delete, and display elements of a list
numbers = [10, 20, 30, 40, 50]

while True:
    print("\n1. Insert")
    print("2. Delete")
    print("3. Display")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        value = int(input("Enter element to insert: "))
        index = int(input("Enter index: "))
        numbers.insert(index, value)
        print("Element inserted.")

    elif choice == 2:
        index = int(input("Enter index to delete: "))
        numbers.pop(index)
        print("Element deleted.")

    elif choice == 3:
        print("List:", numbers)

    elif choice == 4:
        print("Program ended.")
        break

    else:
        print("Invalid choice")
```

```python
#Write a python program to find common elements in two lists
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]

common = []

for value in list1:
    if value in list2:
        common.append(value)

print("List 1:", list1)
print("List 2:", list2)
print("Common elements:", common)
```
