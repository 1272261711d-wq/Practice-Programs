```python
#Write a program to print sum of all the digits of given number
num = int(input("Enter a number: "))

sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10

print("Sum of digits =", sum)
```

```python
#Find the factorial of a given number
num = int(input("Enter a number: "))

factorial = 1

for i in range(1, num+1):
    factorial = factorial*i
print("Factorial=", factorial)
```

```python
#Write a python program to check if a given number is prime or not
num = int(input("Enter a number: "))

if num <= 1:
    print("The number is not prime")

else:
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False

    if prime:
        print("The number is prime")
    else:
        print("The number is not prime")
```

```python
#Write a python program to print prime numbers between 10 to 99
for num in range(10, 100):
    prime = True
    
    for i in range(2, num):
        if num % i == 0:
            prime = False

    if prime:
        print(num)
```

```python
#Write a program to reverse a given number
num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number =", reverse)
```

```python
#Write a program to determine whether a given year is a leap year or not
year = int(input("Enter a year: "))

if year % 400 == 0:
    print(year, "is a Leap Year")
elif year % 100 == 0:
    print(year, "is not a Leap Year")
elif year % 4 == 0:
    print(year, "is a Leap Year")
else:
    print(year, "is not a Leap Year")
```

```python
#Write a python program to accept Marks in Math and display respective grade as an output
marks = int(input("Enter marks in Mathematics: "))

if marks < 0 or marks > 100:
    print("Invalid marks")

elif marks == 100:
    print("Grade: O")

if marks >= 90 and marks != 100:
    print("Grade: O")

elif marks >= 80 and marks <= 89:
    print("Grade: A+")

elif marks >= 70 and marks <= 79:
    print("Grade: A")

elif marks >= 60 and marks <= 69:
    print("Grade: B")

elif marks >= 50 and marks <= 59:
    print("Grade: C")

elif marks >= 40 and marks <= 49:
    print("Grade: P")

elif marks >= 0 and marks < 40:
    print("Grade: F")
```

```python
#Accept Student name, Roll number and Marks of 3 subjects from the user. Calculate the percentage of the marks and display it. Display the subject with the highest and the lowest marks
Name = input("Enter Student Name: ")
Roll_No = int(input("Enter Roll Number: "))

Maths = float(input("Enter Maths marks: "))
Science = float(input("Enter Science marks: "))
English = float(input("Enter English marks: "))

percentage = (Maths + Science + English) / 3

print("\nStudent Name:", Name)
print("Roll Number:", Roll_No)

print("Percentage =", percentage, "%")

if Maths >= Science and Maths >= English:
    print("Subject with Highest Marks: Maths")

elif Science >= Maths and Science >= English:
    print("Subject with Highest Marks: Science")

else:
    print("Subject with Highest Marks: English")


if Maths <= Science and Maths <= English:
    print("Subject with Lowest Marks: Maths")

elif Science <= Maths and Science <= English:
    print("Subject with Lowest Marks: Science")

else:
    print("Subject with Lowest Marks: English")
```
