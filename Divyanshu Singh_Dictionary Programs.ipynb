```python
#Write a program to add key and value to a dictionary
my_dict = {0: 10, 1: 20}
my_dict[2] = 30
print("Dictionary:", my_dict)
```

```python
#Write a program to concatenate the following dictionaries to create a new one
dic1 = {1: 10, 2: 20}
dic2 = {3: 30, 4: 40}
dic3 = {5: 50, 6: 60}
new_dict = {}
new_dict.update(dic1)
new_dict.update(dic2)
new_dict.update(dic3)
print("New Dictionary:", new_dict)
```

```python
#Write a program to check if a given key already exists in a dictionary
my_dict = {101:"Amit", 102:"Rahul", 103:"Priya"}
key = int(input("Enter key to search:"))
if key in my_dict:
    print("Key exists in the dictionary")
else:
    print("Key does not exist in the dictionary")
```

```python
#Write a program to iterate over a dictionary using for loop and print the keys alone, values alone and both keys and values
my_dict = {
    101:"Amit",
    102:"Rahul",
}
print("Keys:")
for key in my_dict:
    print(key)
print("\nValues:")
for value in my_dict.values():
    print(value)
print("\nKeys and Values:")
for key, value in my_dict.items():
    print(key, value)
```

```python
#Write a program to prepare a dictionary where the keys are numbers between 1 and 15(both included) and the values are square of the keys
my_dict = {}
for i in range(1, 16):
 my_dict[i] = i**2
print(my_dict)
```

```python
#Write a program to sum all the values in a dictionary, considering the values will be of int type
my_dict = {"a":10, "b":20, "c":30, "d":40}
total = 0
for value in my_dict.values():
    total = total+value
print("Sum of values:", total)
```

```python
#Write a program to convert a list to dictionary
list1 = ["name", "panel", "rollno"]
list2 = ["ABC", "B", 34]
my_dict = dict(zip(list1, list2))
print("Dictionary:", my_dict)
```

```python
#Write a program to convert dictionary to lists. Make one list of keys and another of vlaues
my_dict = {
    "name":"ABC",
    "panel":"B",
    "rollno":34,
}
keys_list=list(my_dict.keys())
values_list=list(my_dict.values())
print("Key List:", keys_list)
print("Value List:", values_list)
```

```python
#Find the mean of all values of dictionary
my_dict = {
    "marks1":23,
    "marks2":123,
    "marks3":43,
    "marks4":13,
    "marks5":39,
}
total = sum(my_dict.values())
mean = total/len(my_dict)
print("Mean of all values:", mean)
```
