```python
#Use the print function
print("Hello World")

#Assign string to a varible using single quote
a = 'Hello World'
print(a)

#Assign string to a varible using double quote
a = "Hello World"
print(a)

#Take string input
name = input("Enter your name: ")
print("Hello", name)

#Get the character at position
a = "Hello, World!"
print(a[4])

#Repeat a string
a = "Hello " #Adding space inside the quotation marks after Hello in input creates space between each Hello in output
print(a * 3)

#Check the string length
a = "Hello, World!"
print(len(a))
```

```python
#String Indexing starts from 0 if we count from beginning i.e left to right and starts from -1 from the last character i.e right to left
a = "Hello World" #Note that the space between the two words Hello and World will also have an Index
print(a[0])
print(a[1])
print(a[2])
print(a[5])
print(a[-1])
print(a[-4])
```

```python
#Slicing
a = "Hello World!"
#Get the characters from position 1 to 4
print(a[1:4])

#Get the characters from start to position 5
print(a[:5])

#Get the characters from position 2 and all the way to end
print(a[2:])

#Reverse the string
print(a[::-1])
```

```python
a = "Hello, World!"

#Return the string in upper case
print(a.upper())

#Return the string in lower case
print(a.lower())

#Return the string where only the first character is in upper case and the others in lower case
print(a.capitalize())

#Return the string where the first letter of each word is in upper case
print(a.title())

#Return the string where the cases of the characters is swapped
print(a.swapcase())

#Split the string into substrings if it finds instances of the separator using blank space
print(a.split())  #It breaks text wherever it finds any spaces or newlines

#Split the string into substrings if it finds instances of the separator using a specific character
print(a.split(","))  #It breaks text only at the exact character the user chooses like commma or dash
```

```python
#Merge varible a with variable b into variable c using string concatenation (where the user joins the strings directly using '+')
a = "Hello"
b = " World" #Adding space inside the quotation marks before World or after Hello in input creates space between Hello and World in output
c = a + b
print(c)

a = "Hello"
b = "World" #If the space is not added inside quotation marks then Hello and World will become one single word in output
c = a + b
print(c)
```

```python
#Join multiple strings stored in a list together to make a single string using a specified separator
a = ["Hello", "World", "Python"]
print(" ".join(a)) #Here (space, -, comma, /) have been used as separators
print("-".join(a))
print(",".join(a))
print("/".join(a))
```

```python
a = " Hello World "
print(a)

print("\n")

#Remove spaces from both ends of a string
a = " Hello World "
print(a.strip())

print("\n")

#Remove space from the left side of the string
a = " Hello"
print(a)
print(a.lstrip())

print("\n")

#Remove space from the right side of the string
a = "Hello "
print(a)
print(a.rstrip())
```

```python
#Replace characters in a string
a = "I like Physics and Chemistry"
print(a.replace("Chemistry","Maths"))
```

```python
#Find the position\index of the word "Python' in the sentence using find()
a = "I was doing Python assignment yesterday after python class" #The index returned will be of the first occurrence if the same word/character is repeated multiple times in the string
print(a.find("Python"))
print(a.find("Java"))

#Note if the word/character the user wants to find is not there in the string, then find() method will give back the index as -1 but index() method will raise a ValueError and will completely crash the notebook cell
print("\n")

#Find the position\index of the word "Python' in the sentence using index()
a = "I was doing Python assignment yesterday after python class"
print(a.index("doing"))
```

```python
#Count the number of characters in a string
a = "banana"
print(a.count("n"))

print("\n")

#Check the beginning and the ending of the string and return True and False
a = "Hello World"
print(a.startswith("Hello"))
print(a.startswith("World"))
print(a.endswith("Hello"))
print(a.endswith("World"))
```

```python
#Check whether string contains only alphabets
a = "Hello"
print(a.isalpha())

print("\n")

#Check whether string contains only numbers
a = "12345"
print(a.isdigit())

print("\n")

#Check whether string contains only letters and numbers
a = "Hello12345"
print(a.isalnum())

print("\n")

#Check if a string is in lowercase
a = "hello"
print(a.islower())

print("\n")

#Check if a string is in uppercase
a = "HELLO"
print(a.isupper())

print("\n")

#Check for spaces in a string
a = "    "
print(a.isspace())

print("\n")

#Check whether something exists inside a string
a = "Python Programming is easy"
print("Python" in a)
print("Java" in a)

print("\n")

#Compare strings using relational operators such as (==, !=, <, >, <=, >=)
a = "Hello"
b = "Hello"
print(a == b)
print(a != b)
```

```python
#Format the string using f-strings and format() method
Name = "Divyanshu"
Age = 19

print(f"My name is {Name} and my age is {Age}")
print("\n")
print("My name is {} and my age is {}".format(Name, Age))
```

```python
#Check whether a string is a palindrome
a = input("Enter a string: ")

if a == a[::-1]:
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```
