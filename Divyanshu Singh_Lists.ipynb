```python
#Create a list and access the elements
numbers = [10, 20, 30, 40, 50]
print(numbers)
print("\n")

#List can contain diffferent data types
student = ["Divyanshu", 19, 95.5, True]
print(student)
print("\n")

#Access the list elements
words = ["hi", "hello", "bye"]
#(Use positive indexing)
print(words[0]) 
print(words[1])
print(words[2])
print("\n")
#(Use negative indexing)
print(words[-1]) 
print(words[-2])
print(words[-3])
print("\n")

#Access the element of a list which itself is the element of another list
data = ["hi", 19, 95.5, True, [10, 20, 30]]
print(data)
print(data[4][1])
```

```python
#Change the element of a list
words = ["hi", "hello", "bye"] #(Lists are mutable i.e the user can change their elements)
words[1] = "happy"
print(words)
```

```python
#List Slicing [start:stop:step]
words = ["hi", "hello", "bye", "haha", "hoho"]
print(words[1:4]) #(Here the output will show the start index element up to, but not including, the stop index element)
print(words[:2])
print(words[2:])
print(words[:]) #(Here the start and stop index are not specified so python uses default values and copies all the elements of the original list to a new list)
print(words[::-1]) #(Here the list will get reversed)
```

```python
#Adding an element at the end of a list
words = ["hi", "hello", "bye"]
words.append("haha") #(append only adds one item)
print(words)
print("\n")

#(An entire list becomes an element if its added to another list)
words.append([100, 200])
print(words)
```

```python
#Adding multiple elements to the end of a list
words = ["hi", "hello", "bye"]
words.extend([100, 200]) #(extend adds multiple items)
print(words)
```

```python
#Adding an element at a particular index [list.insert(index, element)]
numbers = [10, 20, 30, 40]
numbers.insert(2, 25)
print(numbers)
```

```python
#Removing an element in its first occurence
numbers = [10, 20, 30, 20, 40]
numbers.remove(20)
print(numbers)
```

```python
#Removing an element using its index
numbers = [10, 20, 30, 40]
x = numbers.pop(1) #(Here 1 is the index of the element to be removed)
print(x) #(It will show which element was removed in the output)
print(numbers) #(It will print a new list which won't have the removed element)
```

```python
#Print an empty list
words = []
print(words)
```

```python
#Clearing the elements from the list
numbers = [1, 2, 3, 4, 5]
numbers.clear()
print(numbers)
```

```python
#Tell the index of an element
numbers = [10, 20, 30, 40, 50]
print(numbers.index(30))
```

```python
#Count how many times an element occurs
numbers = [10, 20, 10, 30, 10]
print(numbers.count(10))
```

```python
#Sort the list in order
numbers = [4, 1, 3, 2, 5]
numbers.sort() #(By default python sorts the list in ascending order and sort() changes the original list)
print(numbers)
print("\n")

#Sort in Descending order
numbers.sort(reverse=True)
print(numbers)
print("\n")

#Sort in ascending order
numbers.sort(reverse=False)
print(numbers)
```

```python
#Reverse the existing order of the list
numbers = [10, 20, 30, 40]
numbers.reverse() #(reverse() does not sort the list)
print(numbers)
```

```python
#Create a new sorted list
abc = [4, 1, 3, 2]
new_list = sorted(abc) #(It doesn't change the original list and returns a new list)
print(new_list)
print(abc)
```

```python
#Create a copy of a list
abc = [1, 2, 3]
new_abc = abc.copy() #(Both become separate lists in the output)
print(abc)
print(new_abc)
```

```python
#Join/Merge two lists
a = [1, 2]
b = [3, 4]
print(a + b)
```

```python
#Find common elements
#(Method 1)
a = [1, 2, 3]
b = [4, 3, 5]
common = []
for item in a:
    if item in b:
        common.append(item)
print("Common elements:", common)
print("\n")

#(Method 2)
a = [1, 2, 3]
b = [4, 3, 5]
has_common = bool(set(a) & set(b))
print(has_common)
common = set(a) & set(b)
print(common)
```

```python
#Give the number of elements
numbers = [10, 20, 30, 40, 50]
print(len(numbers))
```

```python
#Check whether an element exists in a list
numbers = [10, 20, 30, 40]
print(20 in numbers)
print(50 in numbers)
print("\n")

#Check whether an element does not exists in a list
numbers = [10, 20, 30]
print(50 not in numbers)
```

```python
#Find the maximum and minimum in a list
numbers = [10, 50, 20, 40, 30]
print("Maximum:", max(numbers))
print("Minimum:", min(numbers))
```

```python
#Find the sum of list elements
numbers = [10, 20, 30, 40]
print("Sum:", sum(numbers))
```
